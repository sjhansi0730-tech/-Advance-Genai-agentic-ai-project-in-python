### 🔑 Key Points about Functions
- Defined using the `def` keyword.
- Can take **parameters** (inputs).
- Can return **values** using `return`.
- Can be called multiple times with different arguments.

### 📌 Types of Functions
1. **Built-in functions** — Already available in Python (e.g., `len()`, `print()`, `sum()`).
2. **User-defined functions** — Functions you create with `def`.
3. **Lambda functions** — Small anonymous functions using `lambda`.

### ⚡ Example: Arithmetic Function
```python
def add_numbers(a, b):
    return a + b

result = add_numbers(5, 7)
print("Sum:", result)
```
**Output:**
```
Sum: 12
```

### ✅ Why Use Functions?
- **Reusability**: Write once, use many times.
- **Modularity**: Break complex problems into smaller parts.
- **Readability**: Code is easier to understand.
- **Maintainability**: Easier to update and debug.

## Variable (*args) And (**kwargs) arguments

#### 🔹 `*args` (Non-keyword variable arguments)
- Collects **extra positional arguments** into a **tuple**.
- Useful when you don’t know how many values will be passed.

#### 🔹 `**kwargs` (Keyword variable arguments)
- Collects **extra keyword arguments** into a **dictionary**.
- Useful when you want flexible named inputs.


````python
def demo(a, b=10, *args, **kwargs):
    print("a:", a)
    print("b:", b)
    print("args:", args)
    print("kwargs:", kwargs)

demo(1, 2, 3, 4, x=5, y=6)
````

````python
def total(*m):
    x = 0
    for i in m:
        x = x + i
    return x

print(total(90,34,56))    
````

````python
def account(name, age, *skill, **social):
    print(f'Name is {name}')
    print(f'Age is {age}')
    for i in skill:
        print(f'skill is {i}')
    for j,k in social.items():    
        print(f'social is {j} {k}')

account('Aditya',24, 'python' ,'sql','c',phone= 123456789,whatsapp= 12348956894)    
````

## Lambda function

#### 🔹 What is a Lambda Function?
- A **lambda function** is an **anonymous function** (no name) defined using the `lambda` keyword.
- Syntax:
  ```python
  lambda arguments: expression
  ```
- It can take any number of arguments but only **one expression** (which is automatically returned).

---

#### 🔹 Key Points
- Lambdas are **single-expression functions**.
- They return values automatically (no `return` keyword needed).
- Best used for **short, simple operations**.
- For complex logic, stick with `def`.

---


````python
# 1. Basic Lambda
squ = lambda x : x * x
print(squ(5))
````

````python
# 2. Multiple Arguments
add = lambda x,y : x + y
print(add(2,5))
````

````python
# 3. With `map()`
nums = [1, 2, 3, 4]
squares = list(map(lambda x: x**2, nums))
print(squares)   # Output: [1, 4, 9, 16]

nums1 = ['1','3','4','6']
t1 = list(map(int , nums1))
print(t1,type(t1[0]))
````

````python
# 4. With `filter()`

nums = [10, 15, 20, 25, 30]
evens = list(filter(lambda x: x % 2 == 0, nums))
print(evens)   # Output: [10, 20, 30]
````

````python
# 5. With `reduce()`
from functools import reduce

nums = [1, 2, 3, 4]
product = reduce(lambda x, y: x * y, nums)
print(product)
````

## 🏠 Local Variables
- Defined inside a function without `global`.
- Exist only during the function’s execution.
- Do **not** affect global variables.

---

## 🌍 Global Variables with `global` keyword
- Explicitly tells Python to use the global variable inside the function.
- Assignments affect the global variable.

---

## 📚 Global Variables with `globals()` function
- Returns a dictionary of all global names.
- You can access or modify global variables dynamically.

```python
x = 10   # global variable

def demo_globals():
    globals()['x'] = 30   # modifies global x via dictionary
    print("Inside function (globals x):", x)

demo_globals()
print("Outside function (globals x):", x)
# Output:
# Inside function (globals x): 30
# Outside function (globals x): 30
```

---

## ⚖️ Comparison Table

| Feature            | Local Variable | `global` keyword | `globals()` function |
|--------------------|----------------|------------------|----------------------|
| **Type**           | Default scope inside function | Statement | Built-in function |
| **Usage**          | Created automatically inside function | Declares intent to use global variable | Accesses/modifies global variables dynamically |
| **Scope control**  | Limited to function | Explicit, static | Dynamic, dictionary-based |
| **Lifetime**       | Exists only during function execution | Exists until program ends | Exists until program ends |
| **Common use case**| Temporary storage for function logic | Assigning to global variable inside a function | Meta-programming, dynamic manipulation of globals |



````python
xi = 10   # global variable
yi = 12

def demo_local():
    xi = 5 # local variable (shadows global x)
    global yi
    yi = 9
    print("x = ",xi)
    print("y = ",yi)

demo_local()

print("="*10)
print("x = ",xi)
print("y = ",yi)
````

### 🔄 Recursion in Python
- **Definition:** A function that calls itself directly or indirectly.
- **Key idea:** Breaks a big problem into smaller subproblems of the same type.
- **Essential parts:**
  1. **Base case** → Stops the recursion (prevents infinite loop).
  2. **Recursive case** → Function calls itself with a smaller/simpler input.

---

### 📘 Example: Factorial
```python
def factorial(n):
    if n == 0:        # base case
        return 1
    else:
        return n * factorial(n-1)  # recursive case

print(factorial(5))  # Output: 120
```

---

### ⚠️ Things to Remember
- Always define a **base case** to avoid infinite recursion.
- Python has a recursion limit (default ~1000 calls).
- Recursion can be elegant but sometimes less efficient than iteration.


````python
def wish():
    print('hello')
    print('hi')

wish()    
````

````python
def wish():
    print('hello')
    print('hi')
    wish()

wish()    
````

````python
import sys
sys.setrecursionlimit(200)
print(sys.getrecursionlimit())
````

````python
def fact(n):
    if n == 0:
        return 1
    return n * fact(n-1)

print(fact(5))    
````


## 🎭 What is a Decorator?
- A **decorator** is a function that takes another function as input and extends or modifies its behavior **without permanently changing it**.
- In Python, decorators are often used for **logging, authentication, timing, caching, validation**, etc.

---

## 🛠 Basic Decorator Example
```python
def my_decorator(func):
    def wrapper():
        print("Before the function runs")
        func()
        print("After the function runs")
    return wrapper

@my_decorator
def say_hello():
    print("Hello, Joy!")

say_hello()
```

**Output:**
```
Before the function runs
Hello, Joy!
After the function runs
```

👉 The `@my_decorator` line is just shorthand for:
```python
say_hello = my_decorator(say_hello)
```

---

## 📦 Decorator with Arguments
Sometimes you want your decorator to accept parameters.

```python
def repeat(n):
    def decorator(func):
        def wrapper(*args, **kwargs):
            for _ in range(n):
                func(*args, **kwargs)
        return wrapper
    return decorator

@repeat(3)
def greet(name):
    print(f"Hello, {name}!")

greet("Joy")
```

**Output:**
```
Hello, Joy!
Hello, Joy!
Hello, Joy!
```

---

## ⚖️ Comparison with Local/Global/Globals
Decorators don’t directly change variable scope like `local`, `global`, or `globals()`.  
Instead, they **wrap functions** to add extra behavior.  

- **Local variable** → Exists only inside a function.  
- **Global variable** → Declared with `global` or modified via `globals()`.  
- **Decorator** → Wraps a function to extend its behavior dynamically.

---

## 🚀 Real-Life Use Cases
- **Logging**: Track when functions are called.
- **Authentication**: Check user permissions before running a function.
- **Timing**: Measure how long a function takes.
- **Caching**: Store results of expensive computations.

Example: Timing decorator
```python
import time

def timer(func):
    def wrapper(*args, **kwargs):
        start = time.time()
        result = func(*args, **kwargs)
        end = time.time()
        print(f"{func.__name__} took {end - start:.4f} seconds")
        return result
    return wrapper

@timer
def slow_function():
    time.sleep(2)
    print("Done!")

slow_function()
```

---

👉 In short:
- **Local/global/globals** → Control variable scope.  
- **Decorators** → Control function behavior.
